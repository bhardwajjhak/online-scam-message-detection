```python
import pandas as pd
import re
import nltk
```

```python
df = pd.read_csv("spam.csv", encoding='latin-1')
```

```python
df = df[['v1', 'v2']]
df.columns = ['label', 'message']
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df['label'] = df['label'].map({'ham': 0, 'spam': 1})
```

```python
def clean_text(text):
    text = text.lower()   # lowercase
    
    text = re.sub(r'http\S+|www\S+', 'url', text)   # remove URLs
    text = re.sub(r'\S+@\S+', 'email', text)        # remove emails
    text = re.sub(r'\d+', ' number ', text)           # replace numbers
    text = re.sub(r'(number\s+)+', 'number ', text)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)         # remove punctuation
    text = re.sub(r'\s+', ' ', text).strip()        # remove extra spaces
    
    return text
```

```python
df['clean_message'] = df['message'].apply(clean_text)
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    return " ".join([word for word in words if word not in stop_words])

df['clean_message'] = df['clean_message'].apply(remove_stopwords)
```

```python
new_english_data = [
    "Your OTP is 7742, please share it to verify your account.",
    "Dear customer, your KYC is pending. Update now to avoid account block.",
    "Congratulations! You have won Rs. 50,000. Claim now.",
    "Work from home! Earn ₹2000 per day, apply now.",
    "You received Rs. 5000, enter your UPI PIN to accept.",
    "Your parcel is on hold, pay Rs. 100 to release it.",
    "I am in trouble, please send me Rs. 5000 urgently."
]
```

```python
new_df = pd.DataFrame(new_english_data, columns=['message'])
new_df['label'] = 1
```

```python
new_df['clean_message'] = new_df['message'].apply(clean_text)
new_df['clean_message'] = new_df['clean_message'].apply(remove_stopwords)
```

```python
df = pd.concat([df, new_df], ignore_index=True)
```

```python
df.drop('message', axis=1, inplace=True)
```

```python
df.to_csv("cleaned_spam.csv", index=False)
```
