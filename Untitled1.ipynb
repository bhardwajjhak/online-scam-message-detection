```python
import pandas as pd
import numpy as np
import seaborn as sns


import matplotlib.pyplot as plt
```

```python
fake = pd.read_csv("fake.csv")
true = pd.read_csv("true.csv")
```

```python
fake['label'] = 0 #Fake news
true['label'] = 1 #True news
```

```python
data = pd.concat([fake,true],axis=0).reset_index(drop=True)
```

```python
import re
import string
def wordopt(text):
    text = text.lower()
    text = re.sub('\[.*?\]', '', text)
    text = re.sub("\\W"," ",text)
    text = re.sub('https?://\S+|www\.\S+', '', text)
    text = re.sub('<.*?>+', '', text)
    text = re.sub('[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub('\n', '', text)
    text = re.sub('\w*\d\w*', '', text)
    return text

data['text'] = data['text'].fillna('').apply(wordopt)
data['title'] = data['title'].fillna('').apply(wordopt)
```

# date

```python
import pandas as pd
data['date'] = pd.to_datetime(data['date'], errors='coerce')
data['year'] = data['date'].dt.year
data['month'] = data['date'].dt.month
data['day'] = data['date'].dt.day
data.drop(['year', 'month', 'day'], axis=1, inplace=True)
data.drop('date', axis=1, inplace=True,errors= 'ignore')
data.head()
```
